```python
from pathlib import Path
from pyspark.sql import SparkSession

DATA_DIR = Path.home() / 'spark-zoomcamp' / 'data'

spark = SparkSession.builder.master('local[*]').appName('zones').getOrCreate()
```

```python
df_zones = spark.read.csv(
    str(Path.home() / 'spark-zoomcamp' / 'taxi_zone_lookup.csv'),
    header=True,
    inferSchema=True,
)
df_zones.printSchema()
```

```python
df_zones.write.parquet(str(DATA_DIR / 'zones'), mode='overwrite')
```
